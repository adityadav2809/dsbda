```python
import pandas as pd

# Load Titanic dataset (replace with the actual path of your dataset)
titanic_df = pd.read_csv('titanic.csv')

# Display the first few rows of the dataset to understand its structure
print(titanic_df.head())

# Group by 'Pclass' (Passenger Class) and calculate summary statistics for 'Age' and 'Fare'
summary_stats = titanic_df.groupby('Pclass')[['Age', 'Fare']].describe()

# Display the summary statistics
print(summary_stats)
```

```python
import pandas as pd
import numpy as np

# Load the dataset (replace with the actual path to your dataset)
df = pd.read_csv('iris.csv')

# Display the first few rows of the dataset to verify its structure
print(df.head())

# Group by 'variety' (species) and calculate basic statistics for the sepal and petal measurements
species_stats = df.groupby('variety').agg(
    mean_sepal_length=('sepal.length', 'mean'),
    std_dev_sepal_length=('sepal.length', 'std'),
    min_sepal_length=('sepal.length', 'min'),
    max_sepal_length=('sepal.length', 'max'),
    percentile_25_sepal_length=('sepal.length', lambda x: np.percentile(x, 25)),
    percentile_50_sepal_length=('sepal.length', lambda x: np.percentile(x, 50)),
    percentile_75_sepal_length=('sepal.length', lambda x: np.percentile(x, 75)),
    
    mean_sepal_width=('sepal.width', 'mean'),
    std_dev_sepal_width=('sepal.width', 'std'),
    min_sepal_width=('sepal.width', 'min'),
    max_sepal_width=('sepal.width', 'max'),
    percentile_25_sepal_width=('sepal.width', lambda x: np.percentile(x, 25)),
    percentile_50_sepal_width=('sepal.width', lambda x: np.percentile(x, 50)),
    percentile_75_sepal_width=('sepal.width', lambda x: np.percentile(x, 75)),
    
    mean_petal_length=('petal.length', 'mean'),
    std_dev_petal_length=('petal.length', 'std'),
    min_petal_length=('petal.length', 'min'),
    max_petal_length=('petal.length', 'max'),
    percentile_25_petal_length=('petal.length', lambda x: np.percentile(x, 25)),
    percentile_50_petal_length=('petal.length', lambda x: np.percentile(x, 50)),
    percentile_75_petal_length=('petal.length', lambda x: np.percentile(x, 75)),
    
    mean_petal_width=('petal.width', 'mean'),
    std_dev_petal_width=('petal.width', 'std'),
    min_petal_width=('petal.width', 'min'),
    max_petal_width=('petal.width', 'max'),
    percentile_25_petal_width=('petal.width', lambda x: np.percentile(x, 25)),
    percentile_50_petal_width=('petal.width', lambda x: np.percentile(x, 50)),
    percentile_75_petal_width=('petal.width', lambda x: np.percentile(x, 75))
)

# Display the statistical details for each variety
print(species_stats)
```
