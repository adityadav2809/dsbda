```python
 # Step 5: Data Formatting and Normalization
# 5.1 Check data types:
df.dtypes
```

```python
# 5.2 Convert data types (if needed):
df['Survived'] = df['Survived'].astype('category')
df['Pclass'] = df['Pclass'].astype('category')
df['Sex'] = df['Sex'].astype('category')
df['Embarked'] = df['Embarked'].astype('category')
```

```python
# 5.3 Normalize numerical values (example on Age and Fare):
df['Age'] = df['Age'].fillna(df['Age'].median())  # Fill missing
df['Fare'] = (df['Fare'] - df['Fare'].mean()) / df['Fare'].std()
df['Cabin'] = df['Cabin'].fillna("Unknown")
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])
df.isnull().sum()
```

```python
# 6. Turning Categorical Variables into Quantitative
#6.1 View Categorical Columns
df.select_dtypes(include='category').columns
```

```python
# 6.2 Apply One-Hot Encoding to the 'Sex', 'Embarked', and 'Pclass' columns
# The drop_first=True argument ensures that we drop the first category of each column to avoid collinearity.
# This is important in some models like Logistic Regression to avoid "dummy variable trap".
df = pd.read_csv("titanic.csv")
df = pd.get_dummies(df, columns=['Sex', 'Embarked', 'Pclass'], drop_first=True)

#  Print the column names of the new dataframe to verify that the categorical columns have been encoded.
print(df.columns)
```

```python
from sklearn.preprocessing import LabelEncoder

# Initialize LabelEncoder
le = LabelEncoder()

# Apply label encoding to the 'Sex_male' column
df['Sex_encoded'] = le.fit_transform(df['Sex_male'])  # 1 for 'male', 0 for 'female'

# Show the first few rows to verify the new 'Sex_encoded' column
print(df.head())
```
